# Examen SQL - Versie E (Abs Bibliotheek)
### Vraag 1 (8 punten)
**1.1 Data Manipulation Language**
Update de tabel `Books` in het `Abs` schema. Verhoog de `duration` (speelduur) van alle boeken die gepubliceerd zijn vóór 2000 (`publishedYear < 2000`) met 10 eenheden.

**1.2 Data Definition Language**
Voeg een constraint toe aan de tabel `Books` zodat de `duration` groter of gelijk aan 0 moet zijn. Noem de constraint `CHK_duration_Books`.

```sql
-- 1.1 Vervanging
UPDATE Abs.Books
SET duration = duration + 10
WHERE publishedYear < 2000;

-- 1.2 Aanbrengen CK
ALTER TABLE Abs.Books
ADD CONSTRAINT CHK_duration_Books CHECK (duration >= 0);
```

### Vraag 2 (8 punten)
Selecteer alle boeken (`Abs.Books`).
Geef in je overzicht:
- Een kolom `BoekInfo` die is opgebouwd uit de titel van het boek en het publicatiejaar tussen haakjes (bijv. "Boektitel (2020)"). Gebruik `ISNULL(CAST(publishedYear AS VARCHAR), 'Onbekend')` indien het jaar niet is ingevuld.
- Een kolom `ToegevoegdOp` gebaseerd op de `createdAt` datum in de format `dd-MM-yyyy`.
- Een kolom `Taal` die de `language` in hoodletters (`UPPER`) weergeeft, en 'ONBEKEND' toont als de taal NULL is.

Sorteer op de originele titel alfabetisch (`title` ASC) en toon enkel de top 10.

**Gewenst resultaat (eerste 10)**

|---|---|---|---|

| 2 | A Brief History of Chronomancy (2025) | 02-04-2025 | ENGLISH |
| 3 | A Conjuring of Light (2017) | 20-05-2024 | ENGLISH |

```sql
-- Antwoord - Vraag 2
SELECT TOP 10 
    CONCAT(title, ' (', ISNULL(CAST(publishedYear AS VARCHAR), 'Onbekend'), ')') AS BoekInfo,
    FORMAT(createdAt, 'dd-MM-yyyy') AS ToegevoegdOp,
    UPPER(ISNULL(language, 'ONBEKEND')) AS Taal
FROM Abs.Books
ORDER BY title ASC;
```

### Vraag 3 (9 punten)
Toon een overzicht van boeken (`Abs.Books`), en hun auteur(s) (`Abs.Authors`). Gebruik de tussentabel `Abs.BookAuthors`.
Toon: de getrimde `title` (`BoekTitel`) en getrimde `name` van de auteur (`AuteurNaam`).
Maak een `DuurCategorie` via een CASE statement op basis van `duration`:
- Kleiner dan 36000: `Kort`
- Tussen 36000 en 72000 (kleiner dan): `Gemiddeld`
- Anders: `Lang`

Sorteer van langste duur naar kortste. Toon de TOP 10.

**Gewenst resultaat (eerste 10)**
| | BoekTitel | AuteurNaam | DuurCategorie |
|---|---|---|---|
| 1 | Wind and Truth | Brandon Sanderson | Lang |
| 2 | Rhythm of War | Brandon Sanderson | Lang |
| 3 | Oathbringer | Brandon Sanderson | Lang |
| 4 | Words of Radiance | Brandon Sanderson | Lang |
| 5 | The Way of Kings | Brandon Sanderson | Lang |
| 6 | The Doors of Midnight | R.R. Virdi | Lang |
| 7 | The Wandering Inn | Pirate Aba | Lang |
| 8 | The Wise Man's Fear | Patrick Rothfuss | Lang |
| 9 | The Navigator's Children | Tad Williams | Lang |
| 10 | Kingdom Blades | Melissa McPhail | Lang |

```sql
-- Antwoord - Vraag 3
SELECT TOP 10
    TRIM(b.title) AS BoekTitel,
    TRIM(a.name) AS AuteurNaam,
    CASE 
        WHEN b.duration < 36000 THEN 'Kort'
        WHEN b.duration < 72000 THEN 'Gemiddeld'
        ELSE 'Lang'
    END AS DuurCategorie
FROM Abs.Books b
INNER JOIN Abs.BookAuthors ba ON b.id = ba.bookId
INNER JOIN Abs.Authors a ON ba.authorId = a.id
ORDER BY b.duration DESC;
```

### Vraag 4 (9 punten)
Overzicht per auteur (`Abs.Authors`): hoeveel boeken (`AantalBoeken`) heeft elke auteur en wat is de totale luistertijd (som van `duration` via `Abs.Books`, aliased als `TotaleDuur`).
Gebruik outer joins zodat auteurs zonder boeken er ook bij staan (toon `TotaleDuur` dan als 0 in plaats van NULL).
Groepeer per auteur (`name` getrimd) en sorteer op `AantalBoeken` aflopend. Toon de top 10.

**Gewenst resultaat (eerste 10)**
| | AuteurNaam | AantalBoeken | TotaleDuur |
|---|---|---|---|
| 1 | Seanan McGuire | 139 | 1769244.8101550001 |
| 2 | Brandon Sanderson | 49 | 3154752.7300000004 |
| 3 | Ursula K. Le Guin | 28 | 257680.53 |
| 4 | Will Wight | 27 | 1109348.3960000002 |
| 5 | Jim Butcher | 25 | 1141375.5930000001 |
| 6 | Robert Jordan | 23 | 2271490.4480000003 |
| 7 | Patricia Briggs | 23 | 810105.799 |
| 8 | Michael J. Sullivan | 20 | 739837.67799999996 |
| 9 | Faith Hunter | 20 | 858350.93000000005 |
| 10 | Brian McClellan | 19 | 353727.83695899998 |

```sql
-- Antwoord - Vraag 4
SELECT TOP 10
    TRIM(a.name) AS AuteurNaam,
    COUNT(b.id) AS AantalBoeken,
    ISNULL(SUM(b.duration), 0) AS TotaleDuur
FROM Abs.Authors a
LEFT JOIN Abs.BookAuthors ba ON a.id = ba.authorId
LEFT JOIN Abs.Books b ON ba.bookId = b.id
GROUP BY a.name
ORDER BY AantalBoeken DESC;
```

### Vraag 5 (8 punten)
Welke auteur (`Abs.Authors`) schreef het absolute langste luisterboek uit de bibliotheek?
Gebruik subqueries via de koppeltabel `Abs.BookAuthors` om deze specifieke auteur te vinden: degene gekoppeld aan het boek wiens `duration` in `Abs.Books` gelijk is aan de `MAX(duration)`.
Zorg voor unieke records met `DISTINCT` en toon de `AuteurNaam`.

**Gewenst resultaat**
| | AuteurNaam |
|---|---|
| 1 | Brandon Sanderson |

```sql
-- Antwoord - Vraag 5
SELECT DISTINCT
    TRIM(a.name) AS AuteurNaam
FROM Abs.Authors a
INNER JOIN Abs.BookAuthors ba ON a.id = ba.authorId
WHERE ba.bookId IN (
    SELECT id 
    FROM Abs.Books 
    WHERE duration = (SELECT MAX(duration) FROM Abs.Books)
);
```

### Vraag 6 (4 punten)
Wat zijn de 5 oudste geregistreerde auteurs in de bibliotheek? (Op basis van de `createdAt` kolom in `Abs.Authors`).
Bereken het aantal dagen tussen hun registratiedatum en 1 augustus 2026 (`'2026-08-01'`) als `DagenSindsAanmaak`.
Sorteer aflopend op dagen.

**Gewenst resultaat (eerste 5)**
| | AuteurNaam | DagenSindsAanmaak |
|---|---|---|
| 1 | Diana Gabaldon | 1256 |
| 2 | Erin Morgenstern | 1256 |
| 3 | Steven Michael Bohls | 1256 |
| 4 | C. L. Clark | 1256 |
| 5 | Django Wexler | 1256 |

```sql
-- Antwoord - Vraag 6
SELECT TOP 5 
    TRIM(name) AS AuteurNaam,
    DATEDIFF(DAY, createdAt, '2026-08-01') AS DagenSindsAanmaak
FROM Abs.Authors
ORDER BY DagenSindsAanmaak DESC;
```
